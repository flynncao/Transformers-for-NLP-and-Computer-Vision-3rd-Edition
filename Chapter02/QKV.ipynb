```python
# QKV
# the formula:
import numpy as np

# head = 8, d_model = 512, n=4
# down to head=2, d_model=8, n = 4, so k=4
# W(Q)= (n, k) = (4,4) ; so as K an V, then h=2, which means in the end we will have 
# b =  2, d_model=8, and n=4 (tokens)

h = 2
d_model  = 8
k = d_model // h
n = 4 # number of tokens

weight_shape = (n, k)

w_q = np.random.randint(1,10, size=weight_shape)
w_k = np.random.randint(1,10, size=weight_shape)
w_v = np.random.randint(1,10, size=weight_shape)

print(w_q)
print(w_k)
print(w_v)

```
